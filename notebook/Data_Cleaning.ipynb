```python
import pandas as pd
import numpy as np

data=pd.read_csv("dataset.csv")
```

### check missing values

```python
#columns contain missing values
data.isnull().sum()[data.isnull().sum() > 0]
```

```python
# % report for missing values
missing_report = pd.DataFrame({
    'Missing Values': data.isnull().sum(),
    'Missing Percentage': (data.isnull().sum() / len(data)) * 100
})

print(missing_report)
```

### check duplicate records

```python
#count duplicate records
data.duplicated().sum()
```

```python
#display duplicate records
data[data.duplicated()]
```

```python
# Remove duplicates
data.drop_duplicates(inplace=True)

# Verify duplicates are removed
print("Duplicates after removal:", data.duplicated().sum())

# Check updated dataset size
print("Updated dataset shape:", data.shape)
```

### check data types

```python
# Check data types
print(data.dtypes)
```

```python
# Find values that cannot be converted into numbers
invalid_values = pd.to_numeric(
    data['writing score'],
    errors='coerce'
)

print(data.loc[invalid_values.isna() & data['writing score'].notna(),
             'writing score'])
```

```python
#print how many times it appear
print((data['writing score'] == 'one hundred').sum())
```

```python
# Correct writing score
data['writing score'] = data['writing score'].replace(
    'one hundred', '100'
)
```

```python
# Convert categorical columns
categorical_cols = [
    'gender',
    'race/ethnicity',
    'parental level of education',
    'lunch',
    'test preparation course'
]

for col in categorical_cols:
    data[col] = data[col].astype('category')


# Convert numerical columns to integer
numerical_cols = [
    'math score',
    'reading score',
    'writing score'
]

for col in numerical_cols:
    data[col] = pd.to_numeric(data[col], errors='coerce').astype('Int64')


# Check updated data types
print(data.dtypes)
```

### check invalid values

```python
# Check minimum and maximum scores
print(data[['math score', 'reading score', 'writing score']].describe())
```

```python
# Replace invalid math scores
data.loc[
    (data['math score'] < 0) | (data['math score'] > 100),
    'math score'
] = np.nan

# Replace invalid reading scores
data.loc[
    (data['reading score'] < 0) | (data['reading score'] > 100),
    'reading score'
] = np.nan
```

```python
#Identify invalid categorical values
categorical_cols = [
    'gender',
    'race/ethnicity',
    'parental level of education',
    'lunch',
    'test preparation course'
]

for col in categorical_cols:
    print(f"\n{col}:")
    print(data[col].unique())
```

```python
#standardize text formatting
categorical_cols = [
    'gender',
    'race/ethnicity',
    'parental level of education',
    'lunch',
    'test preparation course'
]

for col in categorical_cols:
    data[col] = data[col].str.strip().str.lower()
```

```python
#Identify invalid categorical values
categorical_cols = [
    'gender',
    'race/ethnicity',
    'parental level of education',
    'lunch',
    'test preparation course'
]

for col in categorical_cols:
    print(f"\n{col}:")
    print(data[col].unique())
```

```python
#correct spellings
# Correct Gender
data['gender'] = data['gender'].replace({
    'femle': 'female',
    'malee': 'male'
})

# Correct Lunch Type
data['lunch'] = data['lunch'].replace({
    'standrd': 'standard',
    'free reduced': 'free/reduced'
})

# Correct Test Preparation
data['test preparation course'] = data['test preparation course'].replace({
    'completd': 'completed',
    'not completed': 'none'
})

#handle unexpected category
data['race/ethnicity'] = data['race/ethnicity'].replace({
    'group f': pd.NA
})

#Verify corrected categorical values
for col in categorical_cols:
    print(f"\n{col}:")
    print(data[col].value_counts(dropna=False))
```

```python
print(data[categorical_cols].dtypes)
```

```python
#Convert columns back to category data type
for col in categorical_cols:
    data[col] = data[col].astype('category')
```

### rename columns

```python
#rename column names
data.rename(columns={
    'gender': 'Gender',
    'race/ethnicity': 'Race_Ethnicity',
    'parental level of education': 'Parental_Education',
    'lunch': 'Lunch_Type',
    'test preparation course': 'Test_Preparation',
    'math score': 'Math_Score',
    'reading score': 'Reading_Score',
    'writing score': 'Writing_Score'
}, inplace=True)

print(data.columns.tolist())
```

### handle missing values

```python
#fill missing values using Median
score_cols = ['Math_Score', 'Reading_Score', 'Writing_Score']
for col in score_cols:
    data[col] = data[col].fillna(data[col].median())

#fill missing values using Mode
categorical_cols = [
    'Gender',
    'Race_Ethnicity',
    'Parental_Education',
    'Lunch_Type',
    'Test_Preparation'
]
for col in categorical_cols:
    data[col] = data[col].fillna(data[col].mode()[0])
```

### verify cleaned data

```python
#  Missing Values
print("\n--- Missing Values ---")
print(data.isnull().sum())
print("Total Missing Values:", data.isnull().sum().sum())

# Duplicate Records
print("\n--- Duplicate Records ---")
print("Total Duplicates:", data.duplicated().sum())

# Data Types
print("\n--- Data Types ---")
print(data.dtypes)

# Dataset Shape
print("\n--- Dataset Shape ---")
print("Rows:", data.shape[0])
print("Columns:", data.shape[1])
print("Shape:", data.shape)

# Basic Statistics
print("\n--- Numerical Statistics ---")
print(data.describe())

print("\n--- Complete Statistical Summary ---")
print(data.describe(include='all'))
```

### save cleaned dataset

```python
# Save the cleaned dataset
data.to_csv("cleaned_dataset.csv", index=False)

print("\nCleaned dataset saved successfully!")
```
